```python
import torch
import pytorch_lightning as pl
from pathlib import Path

# Path to the checkpoint file
checkpoint_path = "logs/zengchuanlong_geobfn/compete/checkpoints/epoch=339-combined_score=0.789471-validity=0.932000-stable_valid_uniqueness=0.617000-novelty=0.819413.ckpt"

# Check if the checkpoint file exists
if not Path(checkpoint_path).exists():
    print(f"Checkpoint file {checkpoint_path} does not exist!")
else:
    # Load the checkpoint
    checkpoint = torch.load(checkpoint_path, map_location=torch.device('cpu'))
    
    # Print the keys in the checkpoint dictionary to see its structure
    print("Checkpoint keys:", checkpoint.keys())
    
    # If you need to load the model, you'll need the model class
    # model = YourModelClass.load_from_checkpoint(checkpoint_path)
    
    # Or you can just examine the state_dict
    if 'state_dict' in checkpoint:
        print("Model state_dict keys:", list(checkpoint['state_dict'].keys())[:10], "...")
```
